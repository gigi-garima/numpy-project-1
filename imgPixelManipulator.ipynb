```python

import numpy as np
import matplotlib.pyplot as plt


img = np.zeros((240, 240, 3), dtype=np.uint8)


img[:, :, :] = [20, 25, 40]

print("Image shape:", img.shape)
print("Image size:", img.size)
print("Number of dimensions:", img.ndim)
print("Data type:", img.dtype)


img[20:80, 20:90] = [255, 50, 50]


img[100:160, 20:90] = [50, 255, 100]


img[180:220, 20:90] = [50, 100, 255]


img[20:80, 120:200] = [255, 255, 0]


img[100:160, 120:200] = [180, 50, 255]


img[180:200, 120:220] = [0, 255, 255]


img[205:215, 120:220] = [255, 255, 255]



crop = img[10:120, 10:120]


vertical_flip = img[::-1, :, :]


horizontal_flip = img[:, ::-1, :]



red_channel = img[:, :, 0]
green_channel = img[:, :, 1]
blue_channel = img[:, :, 2]

print("Red channel shape:", red_channel.shape)
print("Green channel shape:", green_channel.shape)
print("Blue channel shape:", blue_channel.shape)


img_float = img.astype(np.float32)


bright_img = np.clip(img_float * 1.4, 0, 255)
bright_img = bright_img.astype(np.uint8)


dark_img = np.clip(img_float * 0.5, 0, 255)
dark_img = dark_img.astype(np.uint8)



inverted_img = 255 - img



bright_mask = np.any(img > 200, axis=2)


bright_pixel_count = np.count_nonzero(bright_mask)

print("Number of bright pixels:", bright_pixel_count)



img_uint8 = np.zeros((240, 240, 3), dtype=np.uint8)
img_float64 = np.zeros((240, 240, 3), dtype=np.float64)

print("uint8 memory:", img_uint8.nbytes, "bytes")
print("float64 memory:", img_float64.nbytes, "bytes")



fig, axes = plt.subplots(2, 4, figsize=(14, 8))

images = [
    (img, "Original Image"),
    (crop, "Cropped Image"),
    (vertical_flip, "Vertical Flip"),
    (horizontal_flip, "Horizontal Flip"),
    (bright_img, "Brightened Image"),
    (dark_img, "Darkened Image"),
    (inverted_img, "Inverted Colors"),
    (red_channel, "Red Channel")
]

for ax, (image, title) in zip(axes.flat, images):
    if title == "Red Channel":
        ax.imshow(image, cmap="gray", vmin=0, vmax=255)
    else:
        ax.imshow(image)

    ax.set_title(title)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
