```python
import numpy as np
np.random.seed(7)
```

```python
marks = np.random.normal(65, 12, (500, 6))
marks = marks.clip(0, 100)

print('Number of students:', marks.shape[0])
print('Number of subjects:', marks.shape[1])
print('First 5 students\' marks:')
print(marks[:5])
```

```python
subject_mean = marks.mean(axis=0)
subject_median = np.median(marks, axis=0)
subject_std = marks.std(axis=0)
subject_var = marks.var(axis=0)

print('Average marks in each subject:')
print(subject_mean)
print('\nMedian marks in each subject:')
print(subject_median)
print('\nStandard deviation in each subject:')
print(subject_std)
print('\nVariance in each subject:')
print(subject_var)
```

```python
student_total = marks.sum(axis=1)
student_average = marks.mean(axis=1)

print('First 10 student totals:')
print(student_total[:10])
print('\nFirst 10 student averages:')
print(student_average[:10])
```

```python
z = (marks - marks.mean(axis=0)) / marks.std(axis=0)
outliers = np.where(np.abs(z) > 2)

print('Number of unusual marks flagged:', len(outliers[0]))
print('\nFirst 10 flagged marks:')
for i in range(min(10, len(outliers[0]))):
    student_index = outliers[0][i]
    subject_index = outliers[1][i]
    print('Student', student_index + 1,
          '| Subject', subject_index + 1,
          '| Mark', round(marks[student_index, subject_index], 2),
          '| Z-score', round(z[student_index, subject_index], 2))
```

```python
rank_order = (-student_total).argsort()

print('Top 10 students by total marks:')
for position in range(10):
    student_index = rank_order[position]
    print('Rank:', position + 1,
          '| Student:', student_index + 1,
          '| Total:', round(student_total[student_index], 2),
          '| Average:', round(student_average[student_index], 2))
```

```python
student_results = np.hstack((marks, student_total.reshape(500, 1), student_average.reshape(500, 1)))

print('Shape of final results:', student_results.shape)
print('First 5 rows (6 subjects, total, average):')
print(student_results[:5])
```

```python
column_names = 'Subject1,Subject2,Subject3,Subject4,Subject5,Subject6,Total,Average'
np.savetxt('student_performance.csv', student_results, delimiter=',', header=column_names, comments='', fmt='%.2f')

from google.colab import files
files.download('student_performance.csv')
```
